```python
#Task-1
email="Amit_ml@gmail.edu"
print("count @ = ",email.count("@"))
print("count . = ",email.count("."))
print("user = ",email[:email.index("@")])
print("Domain = ",email[email.index("@")+1:email.index(".")])
if email.endswith(".com"):
 Domain_Ending="Commercial Domain"
elif email.endswith(".edu"):
 Domain_Ending="Educational Domain"
else:
  Domain_Ending="other Domain"
print(Domain_Ending)
```
