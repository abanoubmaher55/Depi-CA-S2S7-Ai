```python
#Task-2
def is_prime(num:int)->bool:
    '''
    check the number is prime or not
    Args:
       num(int):the number to check prime or not
    Returns:
       bool:true if the number prime and false is not prime   
    '''
    if num<2:
        return False
   
    for i in range(2,num):
      if num%i==0:
         return False
    return True  
    
for num in range(2,999):
   if is_prime(num)and is_prime(num+2):
      print(num,num+2)
```

```python
#task-3
def prime_factors(num:int):
    '''
    this function to print the prime factors of a numbers
    Args:
        num(int): the number to find prime factors

    '''
    factor = 2

    while num > 1:
        if num % factor == 0:
            print(factor)
            num = num // factor
        else:
            factor += 1


number = int(input("Enter a number: "))
prime_factors(number)
```

```python
# Task-4
def decimal(num: int) -> str:
    '''
    Convert a decimal number to binary number

    Args:
        num (int): The decimal number to convert the binary number

    Returns:
        str: The binary representation of the number
    '''
    binary = ""
    while num > 0:
        remainder = num % 2
        binary = str(remainder) + binary
        num = num // 2

    return binary


number = int(input("Enter a decimal number: "))
print(decimal(number))
```

```python
#task-5
def is_perfect(num: int) -> bool:
    '''
    Check if a number is a perfect number or not

    Args:
        num (int): The number to check

    Returns:
        bool: True if the number is perfect and False if it is not
    '''
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    return total == num


start = int(input("please enter start: "))
end = int(input("please enter end: "))

for num in range(start, end + 1):
    if is_perfect(num):
        print(num)
```
