```python
class Rectangle:
    _length=5
    width=10

    def area(self):
        return self._length * self.width
r=Rectangle()
print(r._length) 
print(r.width)   
print(r.area())
```

```python
class Parent:
    def __init__(self) -> None:
        self.__value = 10


class Child(Parent):
    def __init__(self) -> None:
        super().__init__()
        self.__value = 20


c = Child()
print(c.__dict__)
```

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):
        return f"my name is {self.name} and my age is {self.age}"
    def __repr__(self)->str:
        return f"my name is {self.name} and my age is {self.age}"
```

```python
s1=student("test",10)
s2=student("test",10)
s1==s2
```
