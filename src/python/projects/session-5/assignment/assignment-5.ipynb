```python
# Task-2

class TwinPrimes:
    '''
    this class to find the twin primes
    '''
    def __init__(self, limit):
        '''
        this constructor to initialize the upper limit
        '''
        self.limit = limit

    def is_prime(self, num: int) -> bool:
        '''
        this method checks whether the number is prime or not

        Args:
            num (int): the number to check

        Returns:
            bool: True if the number is prime,
                  False if the number is not prime
        '''
        if num < 2:
            return False

        for i in range(2, num):
            if num % i == 0:
                return False

        return True

    def twin_prime(self):
        '''
        this method prints the twin primes
        '''
        for num in range(3, self.limit, 2):
            if self.is_prime(num) and self.is_prime(num + 2):
                print(num, num + 2)


twin = TwinPrimes(1000)
twin.twin_prime()
```

```python
#task-3
class PrimeFactors:
    '''
    this class to find the prime factor
    '''
    def __init__(self,num:int):
        '''
         this constructor to initialize the number
        '''
        self.num=num

    def factors(self):
        '''
        this method to print the prime factors of a numbers
        Args:
            num(int): the number to find prime factors

        '''
        num=self.num
        factor = 2

        while num > 1:
            if num % factor == 0:
                print(factor)
                num = num // factor
            else:
                factor += 1


num= int(input("Enter a number: "))
f=PrimeFactors(num)
f.factors()
```

```python
# Task-4
class DecimalTOBinary:
    '''
    this class to convert decimal number to binary number
    '''
    def __init__(self,num:int):
        '''
        this constructor to initialize the number
                
        '''
        self.num=num

    def convert(self):
        '''
        this method Convert a decimal number to binary number
        '''
        num=self.num
        binary = ""
        if num==0:
            return "0"
        while num > 0:
            remainder = num % 2
            binary = str(remainder) + binary
            num = num // 2

        return binary


num= int(input("Enter a decimal number: "))
c=DecimalTOBinary(num)
c.convert()
```

```python
#task-5
class Perfect:
    '''
    this class to find the ferfect number 
    '''
    def __init__(self,start:int,end:int):
        '''
        this constructor to initialize the number and end number
        '''
        self.start=start
        self.end=end

    def is_perfect(self,num:int) -> bool:
        '''
        Check if a number is a perfect number or not

        Args:
            num (int): The number to check

        Returns:
            bool: True if the number is perfect and False if it is not
        '''
        total = 0

        for i in range(1, num):
            if num % i == 0:
                total += i

        return total == num

    def find_perfect_numbers(self):
        '''
        this method Print all perfect numbers
        '''
        for num in range(self.start, self.end + 1):
            if self.is_perfect(num):
                print(num)


start = int(input("please enter start: "))
end = int(input("please enter end: "))
p=Perfect(start,end)
p.find_perfect_numbers()
```
